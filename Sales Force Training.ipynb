```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = [
    100, 150, 50, 100, 130, 120, 100, 85, 70, 150, 150, 120, 50, 100, 100, 140, 90, 150, 50, 90, 120, 
    100, 110, 75, 65
]

df = pd.DataFrame (data, columns=['TransactionAmount'])
df
```

```python
# Measures of Central Tendency

mean_value = df['TransactionAmount'].mean()
median_value = df['TransactionAmount'].median()
mode_value = df['TransactionAmount'].mode()

mean_value, median_value, mode_value
```

Mean = 102.6
Median = 100.0
Mode = 100


```python
# Measures of Variability

range_value = df['TransactionAmount'].max() - df['TransactionAmount'].min()
variance_value = df['TransactionAmount'].var()
std_value = df['TransactionAmount'].std()
min_value = df['TransactionAmount'].min()
max_value = df['TransactionAmount'].max()
Q1 = df ['TransactionAmount'].quantile(0.25)
Q2 = df ['TransactionAmount'].quantile(0.50)
Q3 = df ['TransactionAmount'].quantile(0.75)
range_value, variance_value, std_value, min_value, max_value, Q1, Q2, Q3
```

While the typical transaction value remains around 100, there is substantial variation in sales performance.

```python
# Defining H0 and H1
# H0 : μ = 100
# H1 : μ > 100
# One Sample T-test if a = 0.05

from scipy import stats
t_stat, p_value = stats.ttest_1samp(df['TransactionAmount'], 100)
t_stat, p_value
```

```python
p_value_one_tailed = p_value / 2
p_value_one_tailed
```

p-value = 0.34326424067190586 > 0.05
fail to reject H0
